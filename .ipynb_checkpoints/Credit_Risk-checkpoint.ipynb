#### Data Validation & outlier Handling
.we remove duplicared rows
.we remove rows with an impossible age or work exprerince
.we remove row with zero or negative income and loan amount
.we check if the intrest rate loock realistic   
.we also flag rows where numbers dont add up correctly

```python
df_copy = df.copy()
print(f'Original Rows:{df_copy.shape[0]}')

# Remove Duplicated Rows
df_copy.drop_duplicates(inplace=True)
print(f'Rows after removing duplicates : {df_copy.shape[0]}')

# remove ages belows 18 or above 100
df_copy = df_copy[df_copy['person_age']>=18] 
df_copy = df_copy[df_copy['person_age']<=100] 

# remove employment length greter than age or existing outliers (>60)

df_copy = df_copy[df_copy['person_emp_length'] <= df_copy['person_age']]
df_copy = df_copy[df_copy['person_emp_length'] <= 60]

# .we remove row with zero or negative income and loan amount

df_copy = df_copy[df_copy['loan_amnt'] > 0]

# .we check if the intrest rate loock realistic
print("Loan intrest rate range from (min:5.42 and max : 23.22)")

# .we also flag rows where numbers dont add up correctly

 
```

## Train_Test_Split


```python
x = df_copy.drop('loan_status',axis=1)
y = df_copy['loan_status']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

```python
numeric_cols = ['person_age', 'person_income', 'person_emp_length', 'loan_amnt',
       'loan_int_rate','loan_percent_income','cb_person_cred_hist_length']

categorical_cols = ['person_home_ownership','loan_intent','loan_grade','cb_person_default_on_file']
```

# Handel Class Imbalance
.very few pepole actually defauit in this data
.if we ignore this model may just faver the majority class
.we give more importentance to the minorty class during traing
.this helpes the model actualy learned to spot defaultes

```python
#0 -> no loan (most)
#1 -> yes loan
# y_train.value_counts()
neg,pos = np.bincount(y_train)
scale_weigths = neg/pos
```

```python
scale_weigths
```

```python
## Data Preprocessing -- pipeline & Columns Transformer
```

```python
## Preprocessing for Logistic Regretion
```

```python
# Logistick Regretion : impute + Scaler Numeric, impute + One hote_encodeing
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler,OneHotEncoder


preprocessor_lr = ColumnTransformer(
    transformers=[
        # pipeline for numerical columns
        ('num',Pipeline([
            ('imputer',SimpleImputer(strategy='median')),
            ('scaler',StandardScaler())
        ]),numeric_cols),

        # pipeline for Categorical cols.
        ('cat',Pipeline([
            ('imputer',SimpleImputer(strategy='constant',fill_value='Missing')),
            ('onehot',OneHotEncoder(handle_unknown='ignore'))
        ]),categorical_cols)
    ]
)
```

## preprocessing for XGBoost

.fill missing numeric values                                                                    
.No Scalling needed here ,since tree modles dont need it,                                       
.Convert category columns into number using one-hot encoding

```python
# XGBoost : impute numeric only (No Scaling), impute + One - HotEncoding

preprocessor_xg = ColumnTransformer(
    transformers=[
        # pipeline for numerical columns
        ('num',Pipeline([
            ('imputer',SimpleImputer(strategy='median')),
            ('scaler',StandardScaler())
        ]),numeric_cols),

        # pipeline for Categorical cols.
        ('cat',Pipeline([
            ('imputer',SimpleImputer(strategy='constant',fill_value='Missing')),
            ('onehot',OneHotEncoder(handle_unknown='ignore'))
        ]),categorical_cols)
    ]
)
```

# Evalueate Helper Function
.We build one common function to check any model perforemance.                                  
. It will show accuracy precision recall F1 score and more                                      
.It will also show a confusion matrix and a precision recall curve                

```python
# We have to pass evluate logistck model and Xgbosst model on the basis of their metrics
from sklearn.metrics import accuracy_score,classification_report, precision_score,recall_score,f1_score,confusion_matrix,precision_recall_curve
from sklearn.metrics import confusion_matrix
def evaluet_model(model_name,model,x_test,y_test,threshold=None):
    y_pred_prob = model.predict_proba(x_test)[:, 1]
    

    if threshold:
        y_pred = (y_pred_prob > threshold).astype(int)
    else:
       y_pred = model.predict(x_test) 
    
    

    # metricess
    accuracy_s = accuracy_score(y_test,y_pred) # Overall model's performance
    precision_s = precision_score(y_test,y_pred)
    recall_s = recall_score(y_test,y_pred)
    f1_s = f1_score(y_test,y_pred)


    # Confusion Matrix
    cm = confusion_matrix(y_test, y_pred)
    class_report = classification_report(y_test,y_pred)
    
    print(f'{model_name}Metrics')
    print(f'Best Threshold : {best_threshold}')
    print(f'Accuracy: {accuracy_s:.2f}')
    print(f'precision: {precision_s:.2f}')
    print(f'recall: {recall_s:.2f}')
    print(f'F1 :{f1_s:.2f}')
    print()
    print(f'{model_name} Classification Report:')
    print(class_report)

    #ploting confusion matrix
    sns.heatmap(cm, annot=True,fmt='d')
    plt.title(f'{model_name} Confusion matrix')
    plt.ylabel('Actual Values')
    plt.xlabel('precision_Values')
    plt.show()

    #ploting ROC-AUC curve
    precision, recalls, threshold = precision_recall_curve(y_test,y_pred_prob)
    plt.plot(recalls,precision)
    plt.xlabel('Recall')
    plt.ylabel('precision')
    plt.title(f'{model_name} precision-Recall Curve')
    plt.show()
```

## Stratified Cross_Validation
. one single train_test_split may not be reliabel.                                              
.we split the traing data into five folds instead .                                             
.Each fold keeps the same defult vs repaid ratio                                                .we check the model score across all folds                                                      

```python
from sklearn.model_selection import StratifiedKFold, cross_validate
cv = StratifiedKFold(n_splits=5,shuffle=True,random_state=42)

scoring = {'roc_auc':'roc_auc','accuracy':'accuracy','precision':'precision','recall':'recall','f1':'f1'}
```

```python
# Logistck Regretion
from sklearn.linear_model import LogisticRegression
lr_cv_pipeline = Pipeline([
    ('preprocessor_lr',preprocessor_lr),
    ('classifier',LogisticRegression(max_iter=1000,class_weight='balanced',random_state=42))
])
lr_cv_pipeline
```

```python
from xgboost import XGBClassifier
xgb_cv_Pipeline = Pipeline([
    ('preprocessor',preprocessor_xg),
    ('classifier',XGBClassifier(
        n_estimator=300,max_depth=5,learning_rate=0.1,scale_Pos_weight=scale_weigths,random_state=42
    ))
])
xgb_cv_Pipeline
```

```python
for cv_name, pipe in [('Logistic Regression',lr_cv_pipeline),
                      ('XGBoost',xgb_cv_Pipeline)]:
                      
                      cv_results = cross_validate(pipe,x_train, y_train, cv=cv, scoring=scoring, n_jobs=-1)
                      print(cv_name)
                      print(f'ROC-AUC : {cv_results['test_roc_auc'].mean()}')
                      print(f'Accuracy : {cv_results['test_accuracy'].mean()}')
                      print(f'Precision : {cv_results['test_precision'].mean()}')
                      print(f'Recall : {cv_results['test_recall'].mean()}')
                      print(f'F1 : {cv_results['test_f1'].mean()}')
                      print()
```

## Baseline Model - Logistic Regretion
. We start with a smale model first                                                     
.this become our benchmark to best                                                              
.we train it and check its score on the test data

```python
from sklearn.pipeline import Pipeline
baseline_model = Pipeline([
    ('preprocessor',preprocessor_lr),
    ('Classifier', LogisticRegression(class_weight='balanced'))
    
])
baseline_model.fit(x_train,y_train)
```

# XGBoost Hyperparametwr Tuning
. 

```python
xg_model = Pipeline([
     ('preprocessor',preprocessor_xg),
     ('classifier',XGBClassifier(learning_rate=0.1,scale_pos_weight=scale_weigths,random_state=42
    ))
])
   
xg_model.fit(x_train,y_train)
```

```python
evaluet_model('Baseline Logistic Model',baseline_model,x_test,y_test)
```

```python
from scipy.stats import uniform,randint
```

```python
from sklearn.model_selection import RandomizedSearchCV
param_grid = {
    'classifier__n_estimators':randint(150,600),
    'classifier__max_depth':randint(3,9),
    'classifier__learning_rate':uniform(0.01,0.5),
    'classifier__subsample':uniform(0.6,0.4),
    'classifier__colsample_bytree':uniform(0.6,0.4),
    'classifier__min_child_weight':randint(1,10),
    'classifier__gamma':uniform(0,5)

}
```

```python
random_search = RandomizedSearchCV(
    xg_model,
    param_grid,
    n_iter=150,
    cv=cv,
    scoring='average_precision',
    n_jobs=-1,
    verbose=1
)
random_search.fit(x_train,y_train)
```

```python
print(f'Score after performing Tunning {random_search.best_score_:.2f}')
```

```python
random_search.best_params_
```

## Compare Both Modeles Side By Side

```python
evaluet_model('Logistic Regression',baseline_model,x_test,y_test)
```

```python
## passing the XGBoost model after performing hyperparametr Turing
evaluet_model('XGBoost',random_search,x_test,y_test)
```

```python
# Get the prob of class 1 from xgboost
prob = xg_model.predict_proba(x_test)[:,1]
# for each threshold calculet precision prediction and recall
precision,recalls,threshold = precision_recall_curve(y_test,prob)
# calculate the score for each threshold
f1 = 2*(precision-recalls) / (precision + recalls + 1e-9)
# Get the threshold that produces thje highest f1-score
best = np.argmax(f1[:-1])
best_threshold = threshold[best] 
# Evaluated the next xgboost model using that threshold
evaluet_model('XGBoost',xg_model,x_test,y_test,best_threshold)
```

## Analyzing Falsh Posative False negative

```python
#Get prediction from the best prediction xgbost model on the test set
y_pred = random_search.predict(x_test)
```

```python
# Create a dataframe to easily compare the actual and predicted values
result_df = x_test.copy()
result_df['actual'] = y_test
result_df['predicted'] = y_pred
```

```python
# Identfy Falsh posetive : Model predicted 1 but the actual was 0

falsh_posetive = result_df[(result_df['actual']==0) & (result_df['predicted']==1)]

# Identfy Falsh negative : Model predicted 0 but the actual was 1

falsh_negative = result_df[(result_df['actual']==1) & (result_df['predicted']==0)]
```

```python
print(f'Falsh posetive : {len(falsh_posetive)}')
print(f'Falsh posetive : {len(falsh_negative)}')
```
